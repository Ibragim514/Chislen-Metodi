```python
# Задание №2

def f(x):
    return 2*x**2 - 2*x - 5

# Начальные границы отрезка
a = 2
b = 3
точность = 0.001

# Цикл продолжается пока разница между a и b больше нужной точности
while (b - a) > точность:
    середина = (a + b) / 2
    
    if f(a) * f(середина) < 0:
        b = середина
    else:
        a = середина

print("Корень уравнения:", round(a, 3))
```

```python
# Задание №3

import math

def g(x):
    return math.sqrt((2 * x + 5) / 2)

x_c = 2.0
точность = 10**-6

# Счетчик итераций
шаг = 0

while True:
    x_н = g(x_c)
    шаг += 1
    
    # Если разница между шагами стала ничтожной — нашелся корень
    if abs(x_н - x_c) < точность:
        break
        
    x_c = x_н

print(f"Корень найден за {шаг} шагов: {round(x_н, 6)}")
```

```python
# Задание №4

def f(x):
    return 2*x**2 - 2*x - 5

def df(x):
    # Производная функции
    return 4*x - 2

a = 2.0
b = 3.0
точность = 10**-6

шаг = 0

while abs(b - a) > точность:
    # Метод хорд
    a = a - (f(a) * (b - a)) / (f(b) - f(a))
    
    # Метод касательных
    b = b - f(b) / df(b)
    
    шаг += 1

корень = (a + b) / 2
print(f"Корень найден за {шаг} шага: {round(корень, 6)}")
```
