#### 1 задание

```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
    return 10 - 0.5 * x**2
def f2(x):
    return 2**(-x)

# Подбираем диапазон x, чтобы были видны точки пересечения
x_range = np.linspace(-5, 7, 1000)

y1 = f1(x_range)
y2 = f2(x_range)

plt.plot(x_range, y1, label='f(x) = 10 - 0.5x ** 2')
plt.plot(x_range, y2, label='f(x) = 2 ** -x')

plt.axhline(y=0, color='k', linestyle='-')
plt.axvline(x=0, color='k', linestyle='-')
plt.grid(True)

# Ограничил y для лучшей видимости пересечений
plt.ylim(-5, 20)

plt.title('График f(x) = 10 - 0.5x ** 2 и f(x) = 2 ** -x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

#### 2 задание

```python
from math import *

def f(x):
    result = 10 - 0.5 * x**2 - 2**(-x)
    return result

a = 4
b = 5
eps = 0.001 

while (b - a) / 2 > eps:
    c = (a + b) / 2
    if f(a) * f(c) < 0:
        b = c
    else:
        a = c
    
Ответ = (a + b) / 2
print(f"Корень: {round(Ответ, 3)}")
```

### Практическая работа № 3

```python
from math import *

# Преобразование уравнения к виду x = ϕ(x)
def f(x):
    return sqrt(20 - 2 * 2**(-x))

x1 = 0
x2 = 4.5
count = 1
epselant = 10**(-3)

print(f'0 - {x2}')

while abs(x2 - x1) > epselant:
    x1 = x2
    x2 = f(x1)
    print(f'{count} - {round(x2, 4)}')
    count += 1

print(f'Ответ: {round(x2, 4)}')
```

### Практическая работа № 4

```python
from math import *

def f(x):
    y = 10 - 0.5 * x**2 - 2**(-x)
    return y


xl = float(input('Введите левый x: '))
xr = float(input('Введите правый x: ')) 
eps = float(input('Введите точность (например, 0.0001): '))


d = (xl - xr) / 2
n = 0
c = xl

print(f'n = {n}   xl = {xl}   xr = {xr}')

while abs(d) > eps:
    # Метод Ньютона (касательных) для левой границы
    xl = round(xl - f(xl) / (-xl + 2**(-xl) * log(2)), 6)
    
    # Метод хорд для правой границы
    xr = round((c * f(xr) - xr * f(c)) / (f(xr) - f(c)), 6)
    
    y = (xl + xr) / 2
    d = (xl - xr) / 2
    n += 1
    print(f'n = {n}   xl = {xl}   xr = {xr}')

x = y
print(f'x = {round(x, 3)}   n = {n}')
```
